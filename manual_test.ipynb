```python
# Importerar klasserna som ska testas
from circle_shape import Circle
from rectangle_shape import Rectangle
```

```python
# Testar att både cirkel och rekangel skapas och kan ritas med hjälp av matplotlib
circle1 = Circle(x=0, y=0, radius=1)
rectangle1 = Rectangle(x=0, y=0, width=1, height=1)

circle1.draw()
rectangle1.draw()
```

```python
# Testar ifall jämförelseoperatorerna fungerar
circle2 = Circle(x=1, y=1, radius=1)
circle3 = Circle(radius=3)

print(circle1 == circle2)   # True
print(circle3 > circle1)    # True
print(circle3 >= circle1)   # True

rectangle2 = Rectangle(width=2, height=2)
print(rectangle1 < rectangle2)  # True
print(rectangle1 >= rectangle2) # False
```

```python
#Testar ifall hjälpfunktioner fungerar
print(circle1.is_unit_circle())  # True

print(rectangle1.is_square())    # True
```

```python
# Kontrollerar ifall figurer kan flyttas i y och x led med hjälp av translate()
print(circle1)

circle1.draw()

circle1.translate(5, 3)

circle1.draw()
```
